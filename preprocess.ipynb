```python
df_notna = df[df.Viablity.notna() & df.Smiles.notna()]
df_notna
```

```python
#Duplicados
len(df_notna.Smiles.unique())
```

```python
df_no_dup = df_notna.drop_duplicates(['Smiles'])
df_no_dup
```

```python
selection = ['Smiles','Viablity']
df_selected = df_no_dup[selection]
df_selected
```

```python
#Limpiar smiles
smiles_list = []
for i in df_selected.Smiles.tolist():
  cpd = str(i).split('.')
  cpd_longest = max(cpd, key = len)
  smiles_list.append(cpd_longest)

df_clean_smiles = pd.DataFrame({
    'Viability': df_selected['Viablity'].values,
    'Smiles': smiles_list
})
df_clean_smiles.head()
```

```python
df_clean_smiles.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 6))

# Temporary slice dropping nulls directly for safety
sns.histplot(data=df_clean_smiles.dropna(subset=['Viability']), x='Viability', bins=50, kde=True, color='skyblue')
plt.title('Distribution of Viability Values')
plt.xlabel('Viability (%)')
plt.ylabel('Frequency count')
```

```python
#Since the data already shows a normal distribution I'll keep it the same
df_clean_smiles.to_csv('preprocessed_dataset.csv',  index=False, decimal='.', encoding='utf-8')
```
